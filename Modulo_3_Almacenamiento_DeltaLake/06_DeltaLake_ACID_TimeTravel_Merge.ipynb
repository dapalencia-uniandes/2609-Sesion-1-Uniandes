```python
# Silver: transacciones — primera carga (version 0 del historial Delta)
df_transacciones_silver = (
    spark.table(tbl(SCHEMA_BRONZE, "transacciones"))
    .dropDuplicates(["transaccion_id"])
    .filter(F.col("monto_total") > 0)
)

(
    df_transacciones_silver.write.format("delta")
    .mode("overwrite").option("overwriteSchema", "true")
    .partitionBy("fecha")
    .saveAsTable(tbl(SCHEMA_SILVER, "transacciones"))
)
print(f"[OK] {tbl(SCHEMA_SILVER, 'transacciones')} — {spark.table(tbl(SCHEMA_SILVER,'transacciones')).count():,} registros")
display(spark.sql(f"DESCRIBE HISTORY {tbl(SCHEMA_SILVER, 'transacciones')}").select("version", "timestamp", "operation"))
```

En Delta, <code style="color:#38bdf8;">UPDATE</code> es <strong style="color:#e2e8f0;">atomico</strong>: crea una nueva version del transaction log; los lectores en curso siguen viendo la version anterior hasta que la operacion termina por completo (aislamiento — la "I" de ACID).</td>

```python
# UPDATE atomico — revision manual de fraude sobre transacciones marcadas como Fallida
spark.sql(f"""
    UPDATE {tbl(SCHEMA_SILVER, 'transacciones')}
    SET estado = 'Completada'
    WHERE estado = 'Fallida' AND monto_total < 100000 AND es_fraude = false
""")

print("UPDATE aplicado. Nueva version del historial:")
display(spark.sql(f"DESCRIBE HISTORY {tbl(SCHEMA_SILVER, 'transacciones')}").select("version", "timestamp", "operation", "operationMetrics"))
```

      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;"><strong style="color:#e2e8f0;">Time Travel</strong> tiene dos usos concretos: (1) <em>auditoria</em> — comparar el estado antes/despues de un cambio, exactamente lo que hacemos aqui; (2) <em>reproducibilidad</em> — si un reporte de negocio cito datos de ayer, se puede consultar exactamente esa version. Ambos son imposibles en una tabla SQL tradicional sin CDC configurado aparte.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">SELECT ... FROM tabla VERSION AS OF N</code> consulta la tabla exactamente como estaba en esa version, leyendo del transaction log que numero de archivo pertenecia a cada version — sin duplicar datos. <code style="color:#38bdf8;">RESTORE TABLE</code> (no ejecutado aqui) revertiria todos los cambios de forma igualmente atomica.</td>
    </tr>
  </table>

```python
TABLA = tbl(SCHEMA_SILVER, "transacciones")
hist = spark.sql(f"DESCRIBE HISTORY {TABLA}").collect()
v_antes = hist[-1]["version"]  # version mas antigua (0)
v_actual = hist[0]["version"]  # version actual

print(f"Comparando version {v_antes} (antes del UPDATE) vs version {v_actual} (actual):")

fallidas_antes  = spark.sql(f"SELECT COUNT(*) FROM {TABLA} VERSION AS OF {v_antes} WHERE estado = 'Fallida'").first()[0]
fallidas_actual = spark.sql(f"SELECT COUNT(*) FROM {TABLA} VERSION AS OF {v_actual} WHERE estado = 'Fallida'").first()[0]

print(f"  {'':30} {'Version '+str(v_antes):>15} {'Version '+str(v_actual):>15}")
print(f"  {'Transacciones Fallida':30} {fallidas_antes:>15,} {fallidas_actual:>15,}")
print(f"\n  Registros reclasificados por el UPDATE: {fallidas_antes - fallidas_actual:,}")

print("\nPara revertir el UPDATE completo (rollback), sin necesidad de backups externos:")
print(f"  RESTORE TABLE {TABLA} TO VERSION AS OF {v_antes}")
print("  (no se ejecuta en este demo — afectaria las sesiones siguientes)")
```

  <h2 style="color:#ffffff;font-size:20px;margin:4px 0 0 0;">MERGE INTO: upsert de un lote incremental</h2>
</div>


En un sistema real, las transacciones no llegan todas de una vez: cada dia arriba un lote nuevo, y ademas algunas transacciones de dias anteriores cambian de estado (una reversion, un contracargo). Esto es exactamente lo que resuelve <code>MERGE INTO</code>: **insertar** lo nuevo y **actualizar** lo existente, en una sola operacion atomica.

```python
NUM_NUEVAS = 2000
metodos = ["Tarjeta_Credito", "Tarjeta_Debito", "PSE", "Wallet"]
paises  = ["CO", "MX", "PE", "CL", "AR"]
dispositivos = ["Mobile", "Desktop", "Tablet"]

# Lote de transacciones NUEVAS (simula la llegada del dia siguiente)
df_nuevas = (
    spark.range(NUM_TRANSACCIONES + 1, NUM_TRANSACCIONES + NUM_NUEVAS + 1)
    .withColumn("transaccion_id", F.col("id"))
    .withColumn("cliente_id", (F.col("id") % NUM_CLIENTES + 1).cast(IntegerType()))
    .withColumn("sesion_id",  F.concat(F.lit("SES-"), F.lpad((F.col("id") % NUM_SESIONES + 1).cast(StringType()), 7, "0")))
    .withColumn("fecha_transaccion", F.current_timestamp())
    .withColumn("monto_total",  (F.rand(SEED + 20) * 950000 + 30000).cast(DecimalType(12, 2)))
    .withColumn("metodo_pago",  F.element_at(F.array(*[F.lit(m) for m in metodos]), (F.col("id") % 4 + 1).cast(IntegerType())))
    .withColumn("dispositivo",  F.element_at(F.array(*[F.lit(d) for d in dispositivos]), (F.col("id") % 3 + 1).cast(IntegerType())))
    .withColumn("pais_cliente", F.element_at(F.array(*[F.lit(p) for p in paises]), (F.col("id") % 5 + 1).cast(IntegerType())))
    .withColumn("pais_ip_transaccion", F.col("pais_cliente"))
    .withColumn("es_fraude", F.rand(SEED + 21) < 0.02)
    .withColumn("estado", F.lit("Completada"))
    .withColumn("fecha", F.to_date("fecha_transaccion"))
    .withColumn("ingested_at", F.current_timestamp())
    .withColumn("_source_system", F.lit("PAYMENTS_GATEWAY"))
    .drop("id")
)

# Lote de ACTUALIZACIONES sobre transacciones existentes (contracargos detectados tarde)
df_actualizadas = (
    spark.table(tbl(SCHEMA_SILVER, "transacciones"))
    .filter("estado = 'Completada'")
    .orderBy(F.rand(SEED + 22))
    .limit(500)
    .withColumn("estado", F.lit("Reembolsada"))
    .withColumn("ingested_at", F.current_timestamp())
)

df_incremental = df_nuevas.unionByName(df_actualizadas)
df_incremental.createOrReplaceTempView("transacciones_incremental")
print(f"Lote incremental: {df_incremental.count():,} filas ({NUM_NUEVAS:,} nuevas + 500 actualizadas)")
```

    <span style="color:#38bdf8;font-size:10px;font-weight:700;letter-spacing:1.5px;">ANTES DE EJECUTAR ESTA CELDA</span>

    <tr>
      <td style="width:78px;padding:8px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">QUE HACE</td>
      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Ejecuta <code style="color:#38bdf8;">MERGE INTO</code>: actualiza las filas que ya existen y agrega las que son nuevas, en una sola operacion.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Sin <code style="color:#38bdf8;">MERGE INTO</code>, esto exigiria leer toda la tabla, aplicar la logica en memoria y reescribirla completa con <code style="color:#38bdf8;">overwrite</code> — costoso e innecesariamente riesgoso. <code style="color:#38bdf8;">MERGE INTO</code> solo toca las filas afectadas.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">ON destino.transaccion_id = origen.transaccion_id</code> define la clave de coincidencia; <code style="color:#38bdf8;">WHEN MATCHED THEN UPDATE SET *</code> actualiza todas las columnas de las filas que coinciden; <code style="color:#38bdf8;">WHEN NOT MATCHED THEN INSERT *</code> agrega las que no existian. <code style="color:#38bdf8;">operationMetrics</code> del historial confirma cuantas filas de cada tipo se tocaron.</td>
    </tr>
  </table>
</div>


```python
spark.sql(f"""
    MERGE INTO {tbl(SCHEMA_SILVER, 'transacciones')} AS destino
    USING transacciones_incremental AS origen
    ON destino.transaccion_id = origen.transaccion_id
    WHEN MATCHED THEN UPDATE SET *
    WHEN NOT MATCHED THEN INSERT *
""")

print("MERGE aplicado. Metricas de la operacion:")
metricas = spark.sql(f"DESCRIBE HISTORY {tbl(SCHEMA_SILVER, 'transacciones')}").first()["operationMetrics"]
for k in ["numTargetRowsInserted", "numTargetRowsUpdated", "numTargetRowsMatchedUpdated", "numOutputRows"]:
    if k in metricas:
        print(f"  {k:<28} {metricas[k]}")

total_final = spark.table(tbl(SCHEMA_SILVER, "transacciones")).count()
print(f"\nTotal de transacciones tras el MERGE: {total_final:,}")
```
