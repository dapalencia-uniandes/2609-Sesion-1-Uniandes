Hasta ahora todas las cargas fueron manuales: generamos un DataFrame y lo escribimos. En un sistema real, los archivos **llegan solos** a un almacenamiento (un Volume, un bucket) y alguien tiene que detectarlos e ingerirlos sin reprocesar lo que ya se cargo. Eso es exactamente lo que hace **Auto Loader**.

      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Escribe un archivo JSON con 200 eventos sinteticos directamente en el Volume <code style="color:#38bdf8;">landing</code> creado en el Setup.</td>

      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Simular la llegada de un archivo nuevo desde un sistema externo (un microservicio de checkout, por ejemplo) — sin este archivo no hay nada nuevo que Auto Loader pueda detectar en la celda siguiente.</td>

      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">dbutils.fs.mkdirs</code>/<code style="color:#38bdf8;">dbutils.fs.put</code> son utilidades de Databricks para manipular archivos en Volumes/DBFS directamente desde Python, sin pasar por Spark.</td>

```python
import json, random
from datetime import datetime, timezone

LANDING_PATH = f"/Volumes/{CATALOG_NAME}/{SCHEMA_BRONZE}/{VOLUME_LANDING}/eventos_carrito"
dbutils.fs.mkdirs(LANDING_PATH)

# Simula un archivo nuevo llegando desde un sistema externo (p.ej. un microservicio de checkout)
nuevos_eventos = []
for i in range(200):
    nuevos_eventos.append({
        "evento_id": 9_000_000 + i,
        "sesion_id": f"SES-{random.randint(1, NUM_SESIONES):07d}",
        "cliente_id": random.randint(1, NUM_CLIENTES),
        "producto_id": random.randint(1, 600),
        "tipo_evento": random.choice(["view_product", "add_to_cart", "begin_checkout", "purchase"]),
        "cantidad": random.randint(1, 3),
        "precio_unitario": round(random.uniform(20000, 500000), 2),
        "event_timestamp": datetime.now(timezone.utc).isoformat(),
    })

archivo = f"{LANDING_PATH}/eventos_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
dbutils.fs.put(archivo, "\n".join(json.dumps(r) for r in nuevos_eventos), overwrite=True)
print(f"[OK] Archivo simulado escrito: {archivo}")
print(f"Eventos en el archivo: {len(nuevos_eventos)}")
```

      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">cloudFiles.format("json")</code> mas <code style="color:#38bdf8;">cloudFiles.schemaLocation</code> (dentro del propio Volume, la practica recomendada) detectan archivos nuevos incrementalmente. <code style="color:#38bdf8;">trigger(availableNow=True)</code> procesa todo lo disponible y se detiene solo — convierte un stream continuo en un batch incremental, ideal para un Job programado.</td>
    </tr>
  </table>
</div>


```python
# Auto Loader — lee SOLO archivos nuevos desde la ultima ejecucion (checkpoint)
CHECKPOINT_PATH = f"/Volumes/{CATALOG_NAME}/{SCHEMA_BRONZE}/{VOLUME_LANDING}/_checkpoints/eventos_carrito"

df_autoloader = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.schemaLocation", CHECKPOINT_PATH)
    .load(LANDING_PATH)
)

(
    df_autoloader.writeStream
    .format("delta")
    .option("checkpointLocation", CHECKPOINT_PATH)
    .trigger(availableNow=True)          # procesa lo disponible y se detiene — ideal para Jobs programados
    .toTable(tbl(SCHEMA_BRONZE, "eventos_carrito_streaming"))
)

print(f"[OK] {tbl(SCHEMA_BRONZE, 'eventos_carrito_streaming')} — {spark.table(tbl(SCHEMA_BRONZE, 'eventos_carrito_streaming')).count():,} registros ingeridos")
display(spark.table(tbl(SCHEMA_BRONZE, "eventos_carrito_streaming")).limit(5))
```

  <span style="color:#0d9488;font-size:11px;font-weight:700;letter-spacing:1px;">PATRON DE PRODUCCION</span><br/>
  <span style="color:#cbd5e1;font-size:13px;"><code style="color:#38bdf8;">trigger(availableNow=True)</code> convierte un stream — que normalmente corre indefinidamente — en un <strong style="color:#e2e8f0;">batch incremental</strong>: procesa todo lo que llego desde la ultima vez y se detiene solo. Este patron es el que se usa dentro de un Job programado (Modulo 5): cada ejecucion procesa unicamente lo nuevo, sin necesidad de un cluster corriendo 24/7. **Lakeflow Declarative Pipelines** (antes "Delta Live Tables") llevan el paradigma declarativo — ya visto en el Modulo 2 con DataFrame API vs SQL — un nivel mas arriba: en vez de escribir <em>como</em> orquestar bronze &rarr; silver &rarr; gold (que tabla depende de cual, en que orden, con reintentos), se declara <em>que</em> tabla existe y de donde sale, y el motor de Lakeflow resuelve el grafo de dependencias, la orquestacion y el manejo de errores automaticamente.

Un pipeline declarativo **no se ejecuta como un notebook normal** — se define en un archivo separado y se corre como un recurso de tipo *Pipeline* desde Workflows. Este curso incluye ese archivo junto a este notebook:
