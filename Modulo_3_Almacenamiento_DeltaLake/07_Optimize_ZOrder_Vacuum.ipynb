      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Consulta <code style="color:#38bdf8;">DESCRIBE DETAIL</code> sobre <code style="color:#38bdf8;">silver.transacciones</code> y calcula el tamano medio por archivo.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Cada <code style="color:#38bdf8;">MERGE INTO</code>, <code style="color:#38bdf8;">UPDATE</code> u <code style="color:#38bdf8;">overwrite</code> puede dejar archivos pequenos adicionales, especialmente en particiones con poco volumen. Muchos archivos pequenos significan mas overhead de metadata y mas tareas diminutas al leer.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">DESCRIBE DETAIL</code> expone <code style="color:#38bdf8;">numFiles</code> y <code style="color:#38bdf8;">sizeInBytes</code> sin escanear los datos; dividir uno entre otro da el tamano medio. Un tamano bien por debajo de ~128 MB por archivo es la senal clasica de "small file problem".</td>
    </tr>
  </table>
</div>


```python
TABLA = tbl(SCHEMA_SILVER, "transacciones")
det = spark.sql(f"DESCRIBE DETAIL {TABLA}").first()
print(f"Tabla: {TABLA}")
print(f"  numFiles     : {det['numFiles']}")
print(f"  sizeInBytes  : {det['sizeInBytes']:,}")
print(f"  tamano medio por archivo: {det['sizeInBytes'] / max(det['numFiles'],1) / 1024:.1f} KB")
print("\nUn tamano medio bien por debajo de ~128 MB por archivo es la senal clasica de 'small file problem'.")
```

      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Compactar archivos pequenos en archivos de tamano optimo, y co-localizar filas con <code style="color:#38bdf8;">cliente_id</code> similar en los mismos archivos — asi un filtro o join por cliente puede saltarse archivos enteros (data skipping) en vez de escanear toda la tabla.</td>

      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">OPTIMIZE</code> hace bin-packing de archivos dentro de cada particion; <code style="color:#38bdf8;">ZORDER BY</code> reordena fisicamente los datos por la columna indicada. Es la optimizacion natural para una columna de <strong style="color:#e2e8f0;">alta cardinalidad</strong> que no conviene usar como particion fisica (Modulo 2, Tema 2.6).</td>
    </tr>
  </table>
</div>


```python
resultado = spark.sql(f"OPTIMIZE {TABLA} ZORDER BY (cliente_id)")
display(resultado)

det_despues = spark.sql(f"DESCRIBE DETAIL {TABLA}").first()
print(f"\nnumFiles antes de OPTIMIZE (referencia Tema 3.4) vs despues: {det['numFiles']} -> {det_despues['numFiles']}")
```

Se muestra aqui como referencia — <code style="color:#38bdf8;">silver.transacciones</code> sigue usando <code style="color:#38bdf8;">ZORDER</code> el resto del curso porque es el fundamento conceptual que pide el temario.</td>

```python
spark.sql(f"""
    CREATE OR REPLACE TABLE {tbl(SCHEMA_SILVER, "transacciones_clustered")}
    CLUSTER BY AUTO
    AS SELECT * FROM {TABLA}
""")
print(f"[OK] {tbl(SCHEMA_SILVER, 'transacciones_clustered')} creada con Liquid Clustering (CLUSTER BY AUTO)")
display(spark.sql(f"SHOW TBLPROPERTIES {tbl(SCHEMA_SILVER, 'transacciones_clustered')}").filter("key LIKE '%clusteringColumns%' OR key LIKE '%clusterBy%'"))
```
