```python
# Fijar el contexto por defecto de la sesion — a partir de aqui, nombres sin calificar
# resuelven contra este catalog y schema
spark.sql(f"USE CATALOG {CATALOG_NAME}")
spark.sql(f"USE SCHEMA {SCHEMA_BRONZE}")

print("Contexto actual:", spark.catalog.currentCatalog(), ".", spark.catalog.currentDatabase())
print("\nAhora 'transacciones' resuelve sin calificar, gracias a USE CATALOG / USE SCHEMA:")
display(spark.sql("SELECT COUNT(*) AS total_transacciones FROM transacciones"))
```

**Flujo tipico con Git folders en Databricks:**
1. `Workspace > Git folders > Add` — se clona el repositorio (como este) dentro del workspace.
2. Cada estudiante trabaja en su propia rama (`feature/nombre-apellido`) para no pisar el trabajo de otros.
3. Los cambios se confirman (`commit`) y se sincronizan (`push`) desde el mismo panel del notebook, sin salir de Databricks.
4. El proyecto final (Modulo 6, Sesiones 15-16) se entrega como un Pull Request sobre este mismo repositorio.

      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;">Doble loop: <code style="color:#38bdf8;">SHOW TABLES IN catalog.schema</code> por cada schema, y dentro, <code style="color:#38bdf8;">count()</code> + <code style="color:#38bdf8;">DESCRIBE DETAIL</code> por cada tabla — acumulando totales en un diccionario Python simple.</td>
    </tr>
  </table>
</div>


```python
print("INVENTARIO DE OBJETOS —", CATALOG_NAME)
print("=" * 72)

resumen_total = {"tablas": 0, "registros": 0, "mb": 0.0}
for schema in [SCHEMA_BRONZE, SCHEMA_SILVER, SCHEMA_GOLD]:
    tablas = spark.sql(f"SHOW TABLES IN {CATALOG_NAME}.{schema}").collect()
    print(f"\n{schema.upper()} — {len(tablas)} tabla(s)")
    for t in tablas:
        if t.isTemporary:
            continue
        full = f"{CATALOG_NAME}.{schema}.{t.tableName}"
        cnt = spark.table(full).count()
        det = spark.sql(f"DESCRIBE DETAIL {full}").first()
        mb  = (det["sizeInBytes"] or 0) / 1024 / 1024
        print(f"  {t.tableName:<20} {cnt:>10,} registros  {mb:>8.1f} MB  formato={det['format']}")
        resumen_total["tablas"] += 1
        resumen_total["registros"] += cnt
        resumen_total["mb"] += mb

print("\n" + "=" * 72)
print(f"TOTAL: {resumen_total['tablas']} tablas | {resumen_total['registros']:,} registros | {resumen_total['mb']:.1f} MB")
print("\nSolo BRONZE deberia tener tablas en este punto del curso — Silver y Gold llegan en los Modulos 2, 3 y 5.")
```
