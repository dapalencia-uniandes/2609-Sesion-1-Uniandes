Antes de tocar una sola linea de codigo, es clave entender **por que existe** el Lakehouse.

| Criterio | Data Warehouse | Data Lake | Lakehouse |
|---|---|---|---|
| Tipos de dato | Estructurado (tablas) | Cualquiera (json, imagenes, logs) | Cualquiera, con metadata estructurada |
| Transacciones ACID | Si (nativo) | No | **Si** (Delta Lake) |
| Costo de almacenamiento | Alto (propietario) | Bajo (object storage) | Bajo (object storage) |
| Motor de consulta | Propietario, acoplado al storage | Cualquiera (desacoplado) | Cualquiera (desacoplado) |
| Machine Learning | Dificil (datos ya agregados) | Facil (datos crudos) | Facil (datos crudos + ACID) |
| Gobierno de datos | Fuerte | Debil ("data swamp") | Fuerte (Unity Catalog) |

**La idea central del Lakehouse:** tomar el almacenamiento barato y abierto del Data Lake (archivos Parquet en object storage) y agregarle encima una capa transaccional — **Delta Lake** — que le da las garantias ACID que antes solo tenia un Data Warehouse. Eso es exactamente lo que vamos a construir con nuestras 5 tablas Bronze.
