  <h2 style="color:#ffffff;font-size:20px;margin:4px 0 0 0;">DataFrame API vs Spark SQL: dos sintaxis, un mismo plan</h2>

Todo el codigo Spark, sin importar si se escribe con el **DataFrame API** o con **Spark SQL**, pasa por el mismo optimizador (**Catalyst**) y termina en el mismo plan fisico. Son dos formas de describir *que* resultado se quiere — Spark decide *como* ejecutarlo. Esto es el paradigma **declarativo** que ya se menciono en el Modulo 1.


<div style="box-sizing:border-box;width:100%;background:#0c1a2e;border:1px solid #1e3a5f;border-left:4px solid #2563eb;border-radius:6px;margin:10px 0;overflow:hidden;">

      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Escribe la misma consulta (contar eventos por tipo) dos veces — una con DataFrame API, otra con Spark SQL — e imprime el plan fisico de cada una.</td>

    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Demostrar en la practica que la eleccion entre DataFrame API y SQL es de <strong style="color:#e2e8f0;">estilo</strong>, no de rendimiento: si el plan fisico es identico, ambas se ejecutan exactamente igual. En equipos reales se elige segun quien mantiene el codigo: un analista SQL prefiere <code style="color:#38bdf8;">%sql</code>, un ingeniero de datos suele preferir el DataFrame API porque es mas facil de testear con pytest.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">.explain()</code> imprime el plan fisico que Catalyst genero para ese DataFrame, sin ejecutarlo todavia — comparar ambos planes linea por linea confirma que son el mismo.</td>
    </tr>
  </table>
</div>


```python
# Misma consulta, dos sintaxis

# 1) DataFrame API
df_api = (
    spark.table(tbl(SCHEMA_BRONZE, "eventos_carrito"))
    .groupBy("tipo_evento")
    .agg(F.count("*").alias("eventos"))
    .orderBy(F.desc("eventos"))
)

# 2) Spark SQL
df_sql = spark.sql(f"""
    SELECT tipo_evento, COUNT(*) AS eventos
    FROM {tbl(SCHEMA_BRONZE, "eventos_carrito")}
    GROUP BY tipo_evento
    ORDER BY eventos DESC
""")

print("Plan fisico — DataFrame API:")
df_api.explain()
print("\nPlan fisico — Spark SQL:")
df_sql.explain()

display(df_api)
```

      <td style="width:78px;padding:8px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">QUE HACE</td>
      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Calcula un perfil de calidad de <code style="color:#38bdf8;">eventos_carrito</code> (Bronze): nulos por columna, duplicados por clave natural y valores fuera de rango.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Antes de limpiar, medimos el problema. Este perfil es el insumo que define <strong style="color:#e2e8f0;">que</strong> reglas de limpieza necesita la capa Silver — el mismo tipo de analisis que se haria sobre cualquier tabla Bronze nueva en un proyecto real.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;">Tres tecnicas distintas: <code style="color:#38bdf8;">F.sum(col.isNull().cast(int))</code> cuenta nulos por columna en una sola pasada; <code style="color:#38bdf8;">distinct().count()</code> compara contra el total para detectar duplicados; <code style="color:#38bdf8;">filter()</code> + <code style="color:#38bdf8;">count()</code> mide cuantas filas violan un rango valido.</td>
    </tr>
  </table>
</div>


```python
df_eventos = spark.table(tbl(SCHEMA_BRONZE, "eventos_carrito"))
total = df_eventos.count()

print("PERFIL DE CALIDAD: eventos_carrito (Bronze)")
print("=" * 60)
print(f"Total de registros: {total:,}")

print("\n1. Nulos por columna:")
display(df_eventos.select([
    F.sum(F.col(c).isNull().cast(IntegerType())).alias(c)
    for c in ["sesion_id", "cliente_id", "producto_id", "cantidad", "precio_unitario"]
]))

print("\n2. Duplicados por clave natural (evento_id):")
unicos = df_eventos.select("evento_id").distinct().count()
print(f"   Registros: {total:,}  |  evento_id unicos: {unicos:,}  |  Duplicados: {total - unicos:,}")

print("\n3. Valores fuera de rango:")
n_cant  = df_eventos.filter("cantidad <= 0").count()
n_precio = df_eventos.filter("precio_unitario <= 0").count()
print(f"   cantidad <= 0        : {n_cant:,}")
print(f"   precio_unitario <= 0 : {n_precio:,}")
```

      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">dropDuplicates</code> sobre la clave natural evita doble conteo en el embudo del Modulo 4; los filtros de rango eliminan lecturas fisicamente imposibles detectadas en el perfil de calidad anterior.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">dropDuplicates([clave])</code> + <code style="color:#38bdf8;">filter()</code> encadenados, luego <code style="color:#38bdf8;">write.format("delta").mode("overwrite").saveAsTable()</code>. Todavia no usamos <code style="color:#38bdf8;">MERGE INTO</code> ni <code style="color:#38bdf8;">OPTIMIZE</code> — eso llega en el Modulo 3; por ahora un <code style="color:#38bdf8;">overwrite</code> simple alcanza porque procesamos el historico completo.</td>
    </tr>
  </table>

```python
# Silver: sesiones_web — deduplicada, sin sesiones con duracion invalida
df_sesiones_silver = (
    spark.table(tbl(SCHEMA_BRONZE, "sesiones_web"))
    .dropDuplicates(["sesion_id"])
    .filter((F.col("duracion_seg") > 0) & (F.col("duracion_seg") < 7200))
)

(
    df_sesiones_silver.write.format("delta")
    .mode("overwrite").option("overwriteSchema", "true")
    .partitionBy("fecha")
    .saveAsTable(tbl(SCHEMA_SILVER, "sesiones_web"))
)
print(f"[OK] {tbl(SCHEMA_SILVER, 'sesiones_web')} — {spark.table(tbl(SCHEMA_SILVER,'sesiones_web')).count():,} registros")

# Silver: eventos_carrito — deduplicado, sin cantidades/precios invalidos
df_eventos_silver = (
    spark.table(tbl(SCHEMA_BRONZE, "eventos_carrito"))
    .dropDuplicates(["evento_id"])
    .filter((F.col("cantidad") > 0) & (F.col("precio_unitario") > 0))
)

(
    df_eventos_silver.write.format("delta")
    .mode("overwrite").option("overwriteSchema", "true")
    .partitionBy("fecha")
    .saveAsTable(tbl(SCHEMA_SILVER, "eventos_carrito"))
)
print(f"[OK] {tbl(SCHEMA_SILVER, 'eventos_carrito')} — {spark.table(tbl(SCHEMA_SILVER,'eventos_carrito')).count():,} registros")
```
