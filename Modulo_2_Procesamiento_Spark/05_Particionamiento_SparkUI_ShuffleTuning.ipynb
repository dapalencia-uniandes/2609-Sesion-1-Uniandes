```python
df_eventos  = spark.table(tbl(SCHEMA_SILVER, "eventos_carrito"))
df_productos = spark.table(tbl(SCHEMA_BRONZE, "productos")).select("producto_id", "categoria", "marca")

print("Plan SIN broadcast explicito (Spark puede o no decidir broadcastear solo):")
df_eventos.join(df_productos, on="producto_id").explain()

print("\nPlan CON broadcast explicito:")
df_eventos.join(broadcast(df_productos), on="producto_id").explain()
```

      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><strong style="color:#e2e8f0;">AQE</strong>, activo por defecto en Databricks, ajusta el numero de particiones automaticamente en tiempo de ejecucion fusionando particiones pequenas. Lo activamos explicitamente con <code style="color:#38bdf8;">spark.conf.set()</code> para que el comportamiento sea el mismo sin importar el tipo de compute; el <code style="color:#38bdf8;">try/except</code> cubre el caso en que Serverless ya gestiona ese parametro y no permite sobreescribirlo.</td>
    </tr>
  </table>
</div>


```python
for key, val in [("spark.sql.adaptive.enabled", "true"), ("spark.sql.shuffle.partitions", "auto")]:
    try:
        spark.conf.set(key, val)
        print(f"[OK] {key} = {val}")
    except Exception as e:
        print(f"{key}: gestionado automaticamente por Serverless ({e})")

# Comparacion de particiones antes/despues de un agregado con shuffle
df_grande = spark.table(tbl(SCHEMA_SILVER, "eventos_carrito"))
print(f"\nParticiones de entrada: {df_grande.rdd.getNumPartitions()}")

t0 = time.time()
resultado = df_grande.groupBy("tipo_evento").count().collect()
print(f"Agregacion ejecutada en {time.time()-t0:.2f}s con AQE habilitado")
```

**Buenas practicas de particionamiento fisico** (columnas de tabla, no de shuffle):
- Particionar por una columna de **baja cardinalidad** y muy usada en filtros — en este curso, `fecha`.
- Nunca particionar por una columna de alta cardinalidad (por ejemplo `cliente_id` o `sesion_id`) — genera miles de carpetas diminutas ("small file problem"), el mismo problema que `OPTIMIZE` + `ZORDER` resuelven en el Modulo 3 para columnas de alta cardinalidad.

      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Calcula el resumen diario del embudo (fecha x segmento x canal x tipo_evento) y lo publica como <code style="color:#38bdf8;">silver.resumen_embudo_diario</code>, particionado por fecha.</td>

      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Cerramos el modulo publicando la tabla <strong style="color:#e2e8f0;">Silver</strong> de referencia que va a alimentar el resto del curso — es la tabla que en el Modulo 3 vamos a optimizar con <code style="color:#38bdf8;">OPTIMIZE</code> + <code style="color:#38bdf8;">ZORDER</code>, y que el Modulo 5 va a agregar hasta Gold.</td>

      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;">Mismo patron de join+broadcast del Tema 2.3, agrupando por 4 columnas en vez de 1; <code style="color:#38bdf8;">partitionBy("fecha")</code> aplica la buena practica de particionamiento fisico del Tema 2.6 (baja cardinalidad, usada en filtros).</td>
    </tr>

```python
df_clientes = spark.table(tbl(SCHEMA_BRONZE, "clientes")).select("cliente_id", "segmento")
df_sesiones = spark.table(tbl(SCHEMA_SILVER, "sesiones_web")).select("sesion_id", "canal_origen", "fecha")

df_resumen_embudo = (
    spark.table(tbl(SCHEMA_SILVER, "eventos_carrito"))
    .join(broadcast(df_clientes), on="cliente_id")
    .join(broadcast(df_sesiones), on="sesion_id")
    .groupBy("fecha", "segmento", "canal_origen", "tipo_evento")
    .agg(F.countDistinct("sesion_id").alias("sesiones"))
)

(
    df_resumen_embudo.write.format("delta")
    .mode("overwrite").option("overwriteSchema", "true")
    .partitionBy("fecha")
    .saveAsTable(tbl(SCHEMA_SILVER, "resumen_embudo_diario"))
)

total = spark.table(tbl(SCHEMA_SILVER, "resumen_embudo_diario")).count()
print(f"[OK] {tbl(SCHEMA_SILVER, 'resumen_embudo_diario')} — {total:,} registros")
display(spark.table(tbl(SCHEMA_SILVER, "resumen_embudo_diario")).orderBy(F.desc("fecha")).limit(10))
```
