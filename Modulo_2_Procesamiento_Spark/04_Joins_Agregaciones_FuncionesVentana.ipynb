<div style="box-sizing:border-box;width:100%;background:#0c1a2e;border:1px solid #1e3a5f;border-left:4px solid #2563eb;border-radius:6px;margin:10px 0;overflow:hidden;">

      <td style="width:78px;padding:8px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">QUE HACE</td>
      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Une (join) <code style="color:#38bdf8;">eventos_carrito</code> con <code style="color:#38bdf8;">clientes</code>, <code style="color:#38bdf8;">sesiones_web</code> y <code style="color:#38bdf8;">productos</code> usando <code style="color:#38bdf8;">broadcast()</code> en las tres tablas pequenas.</td>

      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">clientes</code> y <code style="color:#38bdf8;">productos</code> son tablas pequenas comparadas con <code style="color:#38bdf8;">eventos_carrito</code> (millones de filas potenciales en produccion). Enriquecer el clickstream con segmento/ciudad/canal/categoria es lo que permite, en el siguiente tema, medir el embudo <strong style="color:#e2e8f0;">por</strong> esas dimensiones de negocio.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">broadcast(df)</code> le indica a Spark que copie la tabla pequena completa a cada executor en vez de reparticionar ambos lados del join (shuffle) — la optimizacion de join mas usada en pipelines de datos reales cuando un lado es pequeno.</td>
    </tr>
  </table>

```python
df_eventos  = spark.table(tbl(SCHEMA_SILVER, "eventos_carrito"))
df_clientes = spark.table(tbl(SCHEMA_BRONZE, "clientes")).select("cliente_id", "segmento", "ciudad", "pais")
df_sesiones = spark.table(tbl(SCHEMA_SILVER, "sesiones_web")).select("sesion_id", "canal_origen", "dispositivo")
df_productos = spark.table(tbl(SCHEMA_BRONZE, "productos")).select("producto_id", "categoria", "marca", "precio")

df_eventos_enriquecido = (
    df_eventos
    .join(broadcast(df_clientes), on="cliente_id", how="inner")
    .join(broadcast(df_sesiones), on="sesion_id", how="inner")
    .join(broadcast(df_productos), on="producto_id", how="inner")
)

print(f"Eventos enriquecidos: {df_eventos_enriquecido.count():,}")
display(df_eventos_enriquecido.limit(10))
```

      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Calcula 3 tablas de embudo: global, por segmento de cliente y por canal de adquisicion — todas contando <strong style="color:#e2e8f0;">sesiones distintas</strong> por etapa.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Esta es la pregunta de negocio central del curso: <strong style="color:#e2e8f0;">¿en que etapa del embudo se pierden mas usuarios, y varia eso por segmento o por canal?</strong></td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">groupBy(...).agg(F.countDistinct("sesion_id"))</code> — <code style="color:#38bdf8;">countDistinct</code> es la agregacion correcta aqui (no <code style="color:#38bdf8;">count</code>) porque una sesion puede tener varios eventos <code style="color:#38bdf8;">view_product</code> y debe contar una sola vez por etapa.</td>
    </tr>
  </table>

```python
# Embudo global: sesiones distintas que alcanzaron cada etapa
embudo = (
    df_eventos_enriquecido
    .groupBy("tipo_evento")
    .agg(F.countDistinct("sesion_id").alias("sesiones"))
    .orderBy(F.desc("sesiones"))
)
display(embudo)

# Embudo por segmento de cliente — ¿los clientes VIP abandonan menos?
embudo_segmento = (
    df_eventos_enriquecido
    .filter(F.col("tipo_evento").isin("add_to_cart", "purchase"))
    .groupBy("segmento", "tipo_evento")
    .agg(F.countDistinct("sesion_id").alias("sesiones"))
    .orderBy("segmento", "tipo_evento")
)
display(embudo_segmento)

# Embudo por canal de adquisicion — ¿el trafico pagado convierte mejor que el organico?
embudo_canal = (
    df_eventos_enriquecido
    .filter(F.col("tipo_evento").isin("add_to_cart", "purchase"))
    .groupBy("canal_origen", "tipo_evento")
    .agg(F.countDistinct("sesion_id").alias("sesiones"))
    .orderBy("canal_origen", "tipo_evento")
)
display(embudo_canal)
```

<div style="box-sizing:border-box;width:100%;background:#0c1a2e;border:1px solid #1e3a5f;border-left:4px solid #2563eb;border-radius:6px;margin:10px 0;overflow:hidden;">

      <td style="width:78px;padding:8px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">QUE HACE</td>
      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Para cada sesion, calcula si tuvo un evento <code style="color:#38bdf8;">add_to_cart</code> y si tuvo un <code style="color:#38bdf8;">purchase</code>, y deriva la bandera <code style="color:#38bdf8;">carrito_abandonado</code>; luego mide la tasa de abandono por segmento.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Es la definicion exacta de <strong style="color:#e2e8f0;">carrito abandonado</strong> para este curso: agrego al carrito, no compro. Con la tasa por segmento respondemos si los clientes VIP abandonan menos que los Nuevos.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;">Una <strong style="color:#e2e8f0;">funcion de ventana</strong> calcula un valor por fila sin colapsar el resultado como lo hace <code style="color:#38bdf8;">groupBy</code> — util cuando se necesita el detalle y el agregado al mismo tiempo. <code style="color:#38bdf8;">Window.partitionBy("sesion_id")</code> agrupa los eventos de cada sesion sin perder la granularidad de evento; <code style="color:#38bdf8;">F.max(F.when(...))</code> actua como un "OR logico" sobre todos los eventos de la sesion.</td>
    </tr>
  </table>
</div>


```python
# Deteccion de carritos abandonados con funcion de ventana
w_sesion = Window.partitionBy("sesion_id")

df_flags_sesion = (
    df_eventos_enriquecido
    .withColumn("tuvo_agregar", F.max(F.when(F.col("tipo_evento") == "add_to_cart", 1).otherwise(0)).over(w_sesion))
    .withColumn("tuvo_compra",  F.max(F.when(F.col("tipo_evento") == "purchase",   1).otherwise(0)).over(w_sesion))
    .select("sesion_id", "cliente_id", "segmento", "canal_origen", "tuvo_agregar", "tuvo_compra")
    .distinct()
)

df_carritos_abandonados = (
    df_flags_sesion
    .withColumn("carrito_abandonado", (F.col("tuvo_agregar") == 1) & (F.col("tuvo_compra") == 0))
)

total_con_carrito = df_carritos_abandonados.filter("tuvo_agregar = 1").count()
total_abandonado  = df_carritos_abandonados.filter("carrito_abandonado = true").count()
print(f"Sesiones con add_to_cart: {total_con_carrito:,}")
print(f"Carritos abandonados   : {total_abandonado:,} ({total_abandonado/total_con_carrito*100:.1f}%)")

display(
    df_carritos_abandonados.filter("tuvo_agregar = 1")
    .groupBy("segmento")
    .agg(
        F.count("*").alias("total_carritos"),
        F.sum(F.col("carrito_abandonado").cast("int")).alias("abandonados")
    )
    .withColumn("tasa_abandono_pct", F.round(F.col("abandonados") * 100 / F.col("total_carritos"), 1))
    .orderBy(F.desc("tasa_abandono_pct"))
)
```
