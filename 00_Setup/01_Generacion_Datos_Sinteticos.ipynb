```python
# CREATE CATALOG y schemas Medallion + Volume de aterrizaje para Auto Loader
# IF NOT EXISTS hace el script idempotente — se puede re-ejecutar sin errores

spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG_NAME} COMMENT 'Catalog de entrenamiento — Uniandes / Databricks 2026'")

comentarios = {
    SCHEMA_BRONZE: "Ingesta cruda. Datos tal como vienen del sistema origen. Inmutable.",
    SCHEMA_SILVER: "Datos limpios, tipados y validados. Listos para analisis.",
    SCHEMA_GOLD:   "Agregaciones y KPIs. Consumido por BI, alertas y modelos ML."
}
for schema, comentario in comentarios.items():
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG_NAME}.{schema} COMMENT '{comentario}'")
    print(f"[OK] {CATALOG_NAME}.{schema:<8} -> {comentario}")

spark.sql(f"""
    CREATE VOLUME IF NOT EXISTS {tbl(SCHEMA_BRONZE, VOLUME_LANDING)}
    COMMENT 'Zona de aterrizaje para Auto Loader — Modulos 3 y 5'
""")
print(f"\n[OK] Volume creado: {tbl(SCHEMA_BRONZE, VOLUME_LANDING)}")
```

Cada <code style="color:#38bdf8;">withColumn</code> agrega una columna calculada por <strong style="color:#e2e8f0;">expresion vectorizada</strong> (<code style="color:#38bdf8;">F.element_at</code> sobre un array literal para simular categorias, <code style="color:#38bdf8;">F.rand(SEED)</code> para aleatoriedad reproducible).
```python
from pyspark.sql import functions as F
from pyspark.sql.types import *

NUM_PRODUCTOS = 600  # dimension pequena — no depende de un widget

# TABLA 1: clientes — dimension principal, cargada desde el CRM
ciudades    = ["Bogota", "Medellin", "Cali", "Barranquilla", "Bucaramanga", "Cartagena"]
paises      = ["CO", "MX", "PE", "CL", "AR"]
segmentos   = ["Nuevo", "Recurrente", "VIP"]
dispositivos = ["Mobile", "Desktop", "Tablet"]

df_clientes = (
    spark.range(1, NUM_CLIENTES + 1)
    .withColumn("cliente_id", F.col("id").cast(IntegerType()))
    .withColumn("nombre",     F.concat(F.lit("Cliente_"), F.lpad(F.col("id").cast(StringType()), 6, "0")))
    .withColumn("email",      F.concat(F.lit("cliente"), F.col("id").cast(StringType()), F.lit("@correo-demo.com")))
    .withColumn("ciudad",     F.element_at(F.array(*[F.lit(c) for c in ciudades]),    (F.col("id") % 6 + 1).cast(IntegerType())))
    .withColumn("pais",       F.element_at(F.array(*[F.lit(p) for p in paises]),      (F.col("id") % 5 + 1).cast(IntegerType())))
    .withColumn("segmento",   F.element_at(F.array(*[F.lit(s) for s in segmentos]),   (F.col("id") % 3 + 1).cast(IntegerType())))
    .withColumn("dispositivo_preferido", F.element_at(F.array(*[F.lit(d) for d in dispositivos]), (F.col("id") % 3 + 1).cast(IntegerType())))
    .withColumn("fecha_registro", F.date_add(F.lit("2021-01-01").cast(DateType()), (F.col("id") * 3 % 1460).cast(IntegerType())))
    .withColumn("activo",     F.rand(SEED) > 0.05)
    .withColumn("ingested_at", F.current_timestamp())
    .withColumn("_source_system", F.lit("CRM_ECOMMERCE"))
    .drop("id")
)
print(f"Clientes generados: {df_clientes.count():,}")
display(df_clientes)
```

```python
# TABLA 2: productos — dimension de catalogo, cargada desde el ERP de inventario
categorias = ["Electronica", "Hogar", "Moda", "Deportes", "Belleza", "Juguetes"]
marcas     = ["MarcaA", "MarcaB", "MarcaC", "MarcaD", "MarcaE"]

df_productos = (
    spark.range(1, NUM_PRODUCTOS + 1)
    .withColumn("producto_id", F.col("id").cast(IntegerType()))
    .withColumn("nombre",      F.concat(F.lit("Producto_"), F.lpad(F.col("id").cast(StringType()), 4, "0")))
    .withColumn("categoria",   F.element_at(F.array(*[F.lit(c) for c in categorias]), (F.col("id") % 6 + 1).cast(IntegerType())))
    .withColumn("marca",       F.element_at(F.array(*[F.lit(m) for m in marcas]),     (F.col("id") % 5 + 1).cast(IntegerType())))
    .withColumn("precio",      (F.rand(SEED + 1) * 480000 + 20000).cast(DecimalType(10, 2)))
    .withColumn("stock",       (F.rand(SEED + 2) * 500).cast(IntegerType()))
    .withColumn("ingested_at", F.current_timestamp())
    .withColumn("_source_system", F.lit("ERP_INVENTARIO"))
    .drop("id")
)
print(f"Productos generados: {df_productos.count():,}")
display(df_productos)
```

```python
# TABLA 4: eventos_carrito — clickstream. ~3 eventos promedio por sesion.
# La distribucion de tipo_evento simula el embudo real de un e-commerce:
# view_product (55%) -> add_to_cart (25%) -> remove_from_cart (5%) -> begin_checkout (9%) -> purchase (6%)

NUM_EVENTOS = NUM_SESIONES * 3

df_eventos_carrito = (
    spark.range(1, NUM_EVENTOS + 1)
    .withColumn("evento_id",   F.col("id"))
    .withColumn("sesion_id",   F.concat(F.lit("SES-"), F.lpad((F.col("id") % NUM_SESIONES + 1).cast(StringType()), 7, "0")))
    .withColumn("cliente_id",  (F.col("id") % NUM_CLIENTES + 1).cast(IntegerType()))
    .withColumn("producto_id", (F.col("id") % NUM_PRODUCTOS + 1).cast(IntegerType()))
    .withColumn("_r", F.rand(SEED + 5))
    .withColumn("tipo_evento",
        F.when(F.col("_r") < 0.55, F.lit("view_product"))
         .when(F.col("_r") < 0.80, F.lit("add_to_cart"))
         .when(F.col("_r") < 0.85, F.lit("remove_from_cart"))
         .when(F.col("_r") < 0.94, F.lit("begin_checkout"))
         .otherwise(F.lit("purchase")))
    .withColumn("cantidad",         (F.rand(SEED + 6) * 3 + 1).cast(IntegerType()))
    .withColumn("precio_unitario",  (F.rand(SEED + 7) * 480000 + 20000).cast(DecimalType(10, 2)))
    .withColumn("_fecha_base", F.date_add(F.lit("2024-01-01").cast(DateType()), (F.col("id") % 365).cast(IntegerType())))
    .withColumn("event_timestamp",
        (F.col("_fecha_base").cast("timestamp").cast("long") + (F.rand(SEED + 8) * 86400).cast("long")).cast("timestamp"))
    .withColumn("fecha", F.to_date("event_timestamp"))
    .withColumn("ingested_at", F.current_timestamp())
    .withColumn("_source_system", F.lit("WEB_ANALYTICS"))
    .drop("id", "_r", "_fecha_base")
)
print(f"Eventos de carrito generados: {df_eventos_carrito.count():,}")
display(df_eventos_carrito.groupBy("tipo_evento").count().orderBy(F.desc("count")))
```

```python
# TABLA 5: transacciones — pagos, con senal sintetica de fraude para el Modulo 6
metodos = ["Tarjeta_Credito", "Tarjeta_Debito", "PSE", "Wallet"]

df_transacciones = (
    spark.range(1, NUM_TRANSACCIONES + 1)
    .withColumn("transaccion_id", F.col("id"))
    .withColumn("cliente_id", (F.col("id") % NUM_CLIENTES + 1).cast(IntegerType()))
    .withColumn("sesion_id",  F.concat(F.lit("SES-"), F.lpad((F.col("id") % NUM_SESIONES + 1).cast(StringType()), 7, "0")))
    .withColumn("_fecha_base", F.date_add(F.lit("2024-01-01").cast(DateType()), (F.col("id") % 365).cast(IntegerType())))
    .withColumn("fecha_transaccion",
        (F.col("_fecha_base").cast("timestamp").cast("long") + (F.rand(SEED + 9) * 86400).cast("long")).cast("timestamp"))
    .withColumn("monto_total",  (F.rand(SEED + 10) * 950000 + 30000).cast(DecimalType(12, 2)))
    .withColumn("metodo_pago",  F.element_at(F.array(*[F.lit(m) for m in metodos]),      (F.col("id") % 4 + 1).cast(IntegerType())))
    .withColumn("dispositivo",  F.element_at(F.array(*[F.lit(d) for d in dispositivos]), (F.col("id") % 3 + 1).cast(IntegerType())))
    .withColumn("pais_cliente", F.element_at(F.array(*[F.lit(p) for p in paises]),       (F.col("id") % 5 + 1).cast(IntegerType())))
    # pais_ip distinto al pais del cliente en ~8% de los casos -> senal de riesgo
    .withColumn("pais_ip_transaccion",
        F.when(F.rand(SEED + 11) < 0.08,
               F.element_at(F.array(*[F.lit(p) for p in paises]), ((F.col("id") + 2) % 5 + 1).cast(IntegerType())))
         .otherwise(F.col("pais_cliente")))
    .withColumn("_hora", F.hour("fecha_transaccion"))
    .withColumn("_senal_pais",    (F.col("pais_ip_transaccion") != F.col("pais_cliente")).cast(IntegerType()))
    .withColumn("_senal_monto",   (F.col("monto_total") > 700000).cast(IntegerType()))
    .withColumn("_senal_horario", ((F.col("_hora") < 5) | (F.col("_hora") > 23)).cast(IntegerType()))
    .withColumn("_score_riesgo",
        F.col("_senal_pais") * F.lit(0.5) + F.col("_senal_monto") * F.lit(0.3) +
        F.col("_senal_horario") * F.lit(0.2) + F.rand(SEED + 12) * F.lit(0.3))
    .withColumn("es_fraude", F.col("_score_riesgo") > 0.65)
    .withColumn("estado",
        F.when(F.col("es_fraude"), F.lit("Reembolsada"))
         .when(F.rand(SEED + 13) < 0.06, F.lit("Fallida"))
         .otherwise(F.lit("Completada")))
    .withColumn("fecha", F.to_date("fecha_transaccion"))
    .withColumn("ingested_at", F.current_timestamp())
    .withColumn("_source_system", F.lit("PAYMENTS_GATEWAY"))
    .drop("id", "_fecha_base", "_hora", "_senal_pais", "_senal_monto", "_senal_horario", "_score_riesgo")
)
print(f"Transacciones generadas: {df_transacciones.count():,}")
display(df_transacciones.groupBy("es_fraude", "estado").count().orderBy(F.desc("count")))
```

```python
# Funcion de escritura reutilizable — patron DRY (Don't Repeat Yourself)
def escribir_bronze(df, tabla, comment, partition_col=None):
    full_name = tbl(SCHEMA_BRONZE, tabla)
    writer = (
        df.write.format("delta")
          .mode("overwrite")
          .option("overwriteSchema", "true")
          .option("delta.autoOptimize.optimizeWrite", "true")  # archivos de tamano optimo al escribir
    )
    if partition_col:
        writer = writer.partitionBy(partition_col)
    writer.saveAsTable(full_name)
    spark.sql(f"COMMENT ON TABLE {full_name} IS '{comment}'")
    print(f"[OK] {full_name} — {spark.table(full_name).count():,} registros")

escribir_bronze(df_clientes,        "clientes",        "Clientes — dimension principal (CRM)")
escribir_bronze(df_productos,       "productos",       "Catalogo de productos (ERP Inventario)")
escribir_bronze(df_sesiones,        "sesiones_web",     "Sesiones de navegacion (Web Analytics)",  partition_col="fecha")
escribir_bronze(df_eventos_carrito, "eventos_carrito",  "Clickstream del carrito de compras",      partition_col="fecha")
escribir_bronze(df_transacciones,   "transacciones",    "Pagos, con senal sintetica de fraude",    partition_col="fecha")

print("\nSetup completado. Proceda con el Modulo 1 — Sesion 1.")
```
